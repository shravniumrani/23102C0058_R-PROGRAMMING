## Lecture 16 – Logical Operators


```r
x <- 8
(x < 10) || (x < 2)

x <- 18
(x < 10) || (x < 2)
```

```r
x <- c(8,18)
(x < 10) || (x < 2)
(x < 10) | (x < 2)
```

```r
x <- 5
(x < 10) && (x > 2)

x <- 15
(x < 10) && (x > 2)
```

```r
x <- c(8,18)
(x < 10) && (x > 2)
(x < 10) & (x > 2)
```

```r
x <- 1:6
(x > 2) & (x < 5)
x[(x > 2) & (x < 5)]
```

```r
x <- 1:6
(x > 2) | (x < 5)
x[(x > 2) | (x < 5)]
```

```r
x <- TRUE
y <- FALSE
x & y
x | y
!x
xor(x,y)
isTRUE(x)
isFALSE(y)
```

## Lecture 17 – Relational and Logical Operators


```r
x <- 5
Logical1 <- (x > 2)
Logical1
is.logical(Logical1)

Logical2 <- (x < 10)
Logical2
is.logical(Logical2)

Logical3 <- (x != 5)
Logical3
is.logical(Logical3)
```

```r
x <- 5
Logical4 <- (2*x > 11)
Logical4
is.logical(Logical4)

Logical5 <- (3*x < 20)
Logical5
is.logical(Logical5)
```
